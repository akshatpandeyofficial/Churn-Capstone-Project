# Visualization 

```python
df["Monthly Charges"].hist()
plt.show()
```

```python
df.boxplot(column="Monthly Charges")
plt.show()
```

```python
df["Contract"].value_counts().plot(kind="bar")
plt.show()
```

```python
df["Churn Label"].value_counts().plot(kind="bar")
plt.show()
```

```python
df["Churn Label"].value_counts().plot(kind="bar")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python
pd.crosstab(df["Contract"], df["Churn Label"]).plot(kind="bar")
plt.title("Churn by Contract")
plt.show()
```

```python
pd.crosstab(df["Internet Service"], df["Churn Label"]).plot(kind="bar")
plt.title("Churn by Internet Service")
plt.show()
```

```python
pd.crosstab(df["Internet Service"], df["Churn Label"]).plot(kind="bar")
plt.title("Churn by Internet Service")
plt.show()
```

```python
df[df["Churn Label"] == "Yes"]["Churn Reason"].value_counts().plot(kind="bar")
plt.title("Reasons for Customer Churn")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
df["Monthly Charges"].hist()
plt.title("Monthly Charges Distribution")
plt.xlabel("Monthly Charges")
plt.ylabel("Customers")
plt.show()
```

```python
df["Tenure Months"].hist()
plt.title("Tenure Distribution")
plt.xlabel("Tenure Months")
plt.ylabel("Customers")
plt.show()
```
